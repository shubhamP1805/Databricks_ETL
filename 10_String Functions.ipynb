```python
df = spark.read.format('csv')\
    .options(header= True, inferSchema= True,multiline= True)\
    .load("/Volumes/data_qa/citi/raw_file/citi_emp.csv")

display(df)
```

```python
from pyspark.sql.functions import *
```

```python
display(df.select("Emp_Salary",upper("Emp_Name").alias("Emp_Name")))
```

```python
display(df.withColumn("Emp_Name", upper("Emp_Name")))
```

```python
display(df.withColumn("Emp_name",upper("Emp_name")))
```

```python
display(df.withColumn("Emp_name",lower("Emp_name")))
```

```python
display(df.withColumn("emp_name",substring("emp_name",0,3)))
```

```python
display(df.select(substring("Emp_Name",1,3),"Emp_id","Emp_Salary"))
```

```python
display(df.select(length("Emp_Name").alias("Emp_Name")))
```

```python
display(df.select(concat("Emp_Name", lit("_"), "Emp_Salary").alias("Emp_Details")))
```
