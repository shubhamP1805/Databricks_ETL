**What is union in pyspark?** \
union () combines row of two dataframes into a single Dataframe

```python
data1 = [
    (1, 'Alice', 'HR'),
    (2, 'bob', 'IT')
]

columns = ['id', 'Name', 'dept']

df1 = spark.createDataFrame(data1, columns)

display(df1)
```

| **Operation** | **Meaning**                     |
| --------- | --------------------------- |
| UNION ALL | Combine + keep duplicates   |
| UNION     | Combine + remove duplicates |
| INTERSECT | Common rows                 |
| EXCEPT    | Left minus right            |


```python
data2 = [
    (3, 'charlie', 'Finance'),
    (4, 'Devid', 'IT'),
    (2, 'bob', 'IT')  #dup row
]

columns = ['id', 'Name', 'dept']

df2 = spark.createDataFrame(data2, columns)

display(df2)
```

```python
df_union_all = df1.union(df2)
display(df_union_all)
```

```python
df_union = df1.union(df2).distinct()
display(df_union)
```

```python
df_intersect = df1.intersect(df2)
display(df_intersect)
```

```python
df_except = df1.subtract(df2)
display(df_except)
```

```python
df_except_all = df1.exceptAll(df2)
display(df_except_all)
```
