Filter transformation is similar to the WHERE condition.

```python
df = spark.read.format("csv")\
    .options(header = True,inferschema =True,multiline = True)\
    .load("/Volumes/data_qa/citi/raw_file/citi_emp.csv")

display(df)
```

```python
from pyspark.sql.functions import *
display(df.filter(df.Dept == 'Tester'))
```

```python
display(df.filter(df.Emp_Salary>=50000))
```

```python
display(df.filter(df.Emp_Salary<= 50000))
```

```python
display(df.filter((df.Emp_Salary != 50000) & (df.Dept == 'Tester')))
```

```python
display(df.filter(df.Dept.isin("Tester","Dev")))
```

```python
display(df.filter(~df.Dept.isin("Tester","Dev")))
```

```python
display(df.filter(df.Emp_Salary.between(50000,60000)))
```

```python
display(df.filter(~df.Emp_Salary.between(50000,60000)))
```

```python
display(df.filter(df.Emp_Name.like('S%')))
```

```python
display(df.filter(df.Emp_Name.like('%m')))
```

```python
display(df.filter(df.Emp_Name.like('%u%')))
```

**order by**

```python
display(df.orderBy(df.Emp_Salary.desc()))
```

```python
display(df.orderBy("Emp_Salary"))
```
