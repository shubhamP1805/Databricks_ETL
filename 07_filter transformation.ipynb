**Filter transformation** is similar to the WHERE condition.

```python
df = spark.read.format("parquet")\
.options(header = True, inferschema = True, multiline = True)\
.load("/Volumes/data_qa/citi/raw_file/orders.parquet")

display(df)
```

```python
display(df.filter(df.payment_method == "Debit Card"))
```

```python
display(df.filter((df.payment_method == "Credit Card") & (df.status == "Done")))
```

```python
display(df.filter((df.payment_method == "Credit Card") | (df.status == "done")))
```
