```python
df = spark.read.format("parquet")\
.options(hearder = True,inferschema = True ,multiline = True)\
.load("/Volumes/data_qa/citi/raw_file/orders.parquet")
```

The requirement is to add a new column where the value will be order_id + 100. For example, if the order_id is 1, the new value will be 101

```python
from pyspark.sql.functions import *
display(df.withColumn("id",col("order_id")+100))
```

```python
display(df.withColumn("state",lit('abc')))
```

```python
display(df.withColumn("order_id",col("order_id")-1))
```

```python
display(df.withColumnRenamed("order_id","id"))
```
