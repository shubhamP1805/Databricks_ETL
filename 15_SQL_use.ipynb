```sql
%python
df = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .load("/Volumes/data_qa/citi/raw_file/citi_emp.csv")

df.createOrReplaceTempView("citi_emp")


```

```sql
spark.sql("SELECT * FROM citi_emp").show()
```

```sql
select * from citi_emp
```

```sql
SELECT * FROM citi_emp
LIMIT 10;
```

```sql
with cte 
as
(select *,dense_rank() over(partition by dept order by emp_salary desc) as rn from citi_emp)
select * from cte where rn=1
```
