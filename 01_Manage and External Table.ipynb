In a **managed table**, both the metadata and the data are managed and stored by Databricks. When the table is dropped, both the metadata and the underlying data are deleted automatically.

```python
%sql
Drop table data_qa.default.emp_info
```

An **external table** is a table where only the metadata is managed by Databricks, while the actual data is stored in an external storage location. When the table is dropped, only the metadata is deleted and the data remains in the storage.external storage (like ADLS, S3, or DBFS path)

```python
%sql
create table data_qa.default.Emp_info(
  Emp_id int,
  Emp_name string,
  Dept string,
  salary decimal(10,2),
  Location string,
  joining_date date
)
using delta
location 'external Location path'
```
