**What is group by () in pyspark?** \
The group by () transformation groups rows that have the same values in specified column(s). \
After the grouping you must apply aggregate function (like count,sum,avg,min,max) \
similar to group by in sql

```python
df = spark.read.format('csv')\
  .options(header = True , inferschema = True)\
  .load("/Volumes/data_qa/citi/raw_file/citi_emp.csv")
display(df)
```

```python
display(df.groupBy("Dept").count())
```

```python
display(df.groupBy("Dept").agg(sum("Emp_Salary").alias("Sum_salary")))
```

```python
from pyspark.sql.functions import *

display(df.groupBy("Dept").agg(sum("Emp_Salary").alias("Sum_salary")))
```

```python
display(df.groupBy("Dept").agg(min("Emp_Salary").alias("Min_salary")))
```

```python
data = [
  (1,'Alice',23,'Tester'),
  (2,'Bob',34,'Tester'),
  (3,'Charlie',29,'Dev'),
  (1,'Alice',23,'Tester'),
  (2,'Bob',34,'Tester'),
  (6,'Alice',23,'Dev')
]
columns = ["id","name","age","Dept"]

df = spark.createDataFrame(data,columns)
display(df)
```

```python
display(df.groupBy("Dept").agg(count("name")))
```
